```python
import pandas as pd

CleanedStudentSuccess = pd.read_csv(r"cleaned_data.csv")
```

```python
list(CleanedStudentSuccess.columns)
```

```python
CleanedStudentSuccess = CleanedStudentSuccess[CleanedStudentSuccess['placement_binary'] == 1]
print("Filtered to placed students only:", CleanedStudentSuccess.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error


train_df, test_df = train_test_split(
    CleanedStudentSuccess,
    test_size=0.2,
    random_state=42,
    shuffle=True   
)

#randomly assign to test and train to improve prediction
```

# Linear Modeling

```python
import statsmodels.formula.api as smf

model1 = smf.ols('salary_lpa ~ internships_completed', data=train_df).fit()

print(model1.summary())
```

```python
model2 = smf.ols('salary_lpa ~ cgpa + internships_completed + resume_score', data=train_df).fit()

print(model2.summary())
```

```python
model3 = smf.ols('salary_lpa ~ cgpa + internships_completed + resume_score + aptitude_score + mock_interview_score + communication_skills + development_projects_count', data=train_df).fit()

print(model3.summary())
```

```python
model4 = smf.ols('salary_lpa ~ cgpa + internships_completed + resume_score+ aptitude_score + communication_skills+ mock_interview_score+ DSA_problems_solved+ development_projects_count+ GitHub_repos+ AI_ML_projects+ interview_rounds_cleared+ offer_count+ adaptability_score', data=train_df).fit()

print(model4.summary())
```

```python
model5 = smf.ols('salary_lpa ~ cgpa + internships_completed + resume_score + aptitude_score + communication_skills + mock_interview_score + DSA_problems_solved + development_projects_count + GitHub_repos + hackathons_participated + AI_ML_projects + interview_rounds_cleared + offer_count + motivation_level + self_learning_hours + college_tier + family_income_lpa + adaptability_score', data=train_df).fit()

print(model5.summary())
```

```python
model6 = smf.ols('salary_lpa ~ cgpa + internships_completed + resume_score + communication_skills + DSA_problems_solved + GitHub_repos + hackathons_participated + AI_ML_projects + interview_rounds_cleared + offer_count + college_tier ', data=train_df).fit()

print(model6.summary())
```

# Evaluating the linear models

##### -Lower RMSE is better

```python
model1_cols = ['salary_lpa', 'internships_completed']
rmse_df_model1 = test_df[model1_cols].dropna()


actual_values_model1 = rmse_df_model1['salary_lpa']
predicted_values_model1 = model1.predict(rmse_df_model1)
rmse_model1 = root_mean_squared_error(actual_values_model1, predicted_values_model1)
print(f"RMSE for Model 1: {rmse_model1}")
```

```python
model2_cols = ['salary_lpa', 'cgpa', 'internships_completed', 'resume_score']
rmse_df_model2 = test_df[model2_cols].dropna()


actual_values_model2 = rmse_df_model2['salary_lpa']
predicted_values_model2 = model2.predict(rmse_df_model2)
rmse_model2 = root_mean_squared_error(actual_values_model2, predicted_values_model2)
print(f"RMSE for Model 2: {rmse_model2}")
```

```python
model3_cols = ['salary_lpa', 'cgpa', 'internships_completed', 'resume_score', 'aptitude_score', 'mock_interview_score', 'communication_skills', 'development_projects_count']
rmse_df_model3 = test_df[model3_cols].dropna()


actual_values_model3 = rmse_df_model3['salary_lpa']
predicted_values_model3 = model3.predict(rmse_df_model3)
rmse_model3 = root_mean_squared_error(actual_values_model3, predicted_values_model3)
print(f"RMSE for Model 3: {rmse_model3}")
```

```python
model4_cols = ['salary_lpa','cgpa','internships_completed','resume_score','aptitude_score','communication_skills','mock_interview_score','DSA_problems_solved','development_projects_count','GitHub_repos','AI_ML_projects','interview_rounds_cleared','offer_count','adaptability_score']
rmse_df_model4 = test_df[model4_cols].dropna()


actual_values_model4 = rmse_df_model4['salary_lpa']
predicted_values_model4 = model4.predict(rmse_df_model4)
rmse_model4 = root_mean_squared_error(actual_values_model4, predicted_values_model4)
print(f"RMSE for Model 4: {rmse_model4}")
```

```python
model5_cols = ['salary_lpa','cgpa','internships_completed','resume_score','aptitude_score','communication_skills','mock_interview_score','DSA_problems_solved','development_projects_count','GitHub_repos','hackathons_participated','AI_ML_projects','interview_rounds_cleared','offer_count','motivation_level','self_learning_hours','college_tier','family_income_lpa','adaptability_score']
rmse_df_model5 = test_df[model5_cols].dropna()


actual_values_model5 = rmse_df_model5['salary_lpa']
predicted_values_model5 = model5.predict(rmse_df_model5)
rmse_model5 = root_mean_squared_error(actual_values_model5, predicted_values_model5)
print(f"RMSE for Model 5: {rmse_model5}")
```

```python
model6_cols = ['salary_lpa','cgpa','internships_completed', 'resume_score', 'communication_skills', 'DSA_problems_solved', 'GitHub_repos', 'hackathons_participated', 'AI_ML_projects', 'interview_rounds_cleared', 'offer_count', 'college_tier']
rmse_df_model6 = test_df[model6_cols].dropna()


actual_values_model6 = rmse_df_model6['salary_lpa']
predicted_values_model6 = model6.predict(rmse_df_model6)
rmse_model6 = root_mean_squared_error(actual_values_model6, predicted_values_model6)
print(f"RMSE for Model 6: {rmse_model6}")
```

From this we with the R squared value and RMSE we can determmine that model6 is the best version. 

```python
import matplotlib.pyplot as plt
import pandas as pd

# Pull coefficients from model6, drop the Intercept (not a "factor")
coefs = model6.params.drop('Intercept')

top10 = coefs.reindex(coefs.abs().sort_values(ascending=False).head(10).index)

plt.figure(figsize=(8,5))
colors = ['green' if v > 0 else 'red' for v in top10]
top10.sort_values().plot(kind='barh', color=colors)
plt.title('Top 10 Factors for Salary (Linear Regression - Model 6)')
plt.xlabel('Coefficient (green = increases salary, red = decreases)')
plt.tight_layout()
plt.savefig('salary_importance_linreg.png', dpi=150)
plt.show()

print(top10.sort_values(ascending=False))
```
